```python
import numpy as np
x = np.array([4])
y = np.array([15])
m = len(x)
w = 2
b = 5
y_predict = w * x + b
print(f"{y_predict}min")
```

```python
erreur = y_predict - y
print(f"L'erreur est : {erreur}")
```

```python
mse = np.sum(erreur ** 2) / m
rmse = np.sqrt(mse)
print(f"l'erreur quadratique moyenne est : {mse}")
print(f"la racine de l'erreur quadratique moyenne est : {rmse}")
```
